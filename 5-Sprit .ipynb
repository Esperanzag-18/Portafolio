[Ahora que los datos están limpios, agrega los datos por usuario y por periodo para que solo haya un registro por usuario y por periodo. Esto facilitará mucho el análisis posterior.]

```python
# Calcula el número de llamadas hechas por cada usuario al mes. Guarda el resultado.
calls["year_month"]= calls["call_date"].dt.to_period("M")
calls_per_user_month=calls.groupby(["user_id","year_month"]).size().reset_index(name="calls_count")
print("numero de llamadas hechas por cada usuario ")
print (calls_per_user_month.head(10))
print(f"Total llamas : {len(calls_per_user_month)}")
```

```python

# Calcula la cantidad de minutos usados por cada usuario al mes. Guarda el resultado.

calls_per_user_minute_month=calls.groupby(["user_id","year_month"])["duration"].sum().reset_index(name="minutes_used")
calls_per_user_minute_month["minutes_used"] = calls_per_user_minute_month["minutes_used"].apply(math.ceil)
calls['duration_rounded'] = calls['duration'].apply(math.ceil)
calls_per_user_minute_month = calls.groupby(['user_id', 'year_month'])['duration_rounded'].sum().reset_index(name='minutes_used')
calls['duration_rounded'] = calls['duration'].apply(math.ceil)
calls_per_user_minute_month = calls.groupby(['user_id', 'year_month'])['duration_rounded'].sum().reset_index(name='minutes_used')
print("cantidad de minutos usados por cada usuario al mes")
print(calls_per_user_minute_month.head())
print(f"total minutos usados: {len(calls_per_user_minute_month)}")
print("cantidad de minutos usados por cada usuario al mes")
print(calls_per_user_minute_month.head())
print(f"total minutos usados :{len(calls_per_user_minute_month)}")
    

```

```python
# Calcula el número de mensajes enviados por cada usuario al mes. Guarda el resultado.
messages["year_month"]=messages["message_date"].dt.to_period("M")
messages_per_user_month = messages.groupby(["user_id","year_month"])["message_date"].size().reset_index(name="message_count")
print("cantidad de mensajes enviados por usuario al mes")
print(messages_per_user_month.head())
print(f"total mensajes:{len(messages_per_user_month)}")
```

```python

# Calcula el volumen del tráfico de Internet usado por cada usuario al mes. Guarda el resultado.
internet["year_month"]=internet["session_date"].dt.to_period("M")
internet_monthly = internet.groupby(['user_id', 'year_month'])['mb_used'].sum().reset_index(name='mb_total')
internet_monthly['gb_used'] = internet_monthly['mb_total'].apply(lambda x: math.ceil(x / 1024))
print("Volumen de trafico de internet por usuario")
print("Internet correctamente redondeado:")
print(internet_monthly.head())
print(f"total volumen usado:{len(internet_monthly)}")
```

```python

# Calcula el ingreso mensual para cada usuario
Total_Fusion = Total_Fusion.fillna(0)

Total_Fusion['excess_minutes'] = (Total_Fusion['minutes_used'] - Total_Fusion['minutes_included']).clip(lower=0)
Total_Fusion['excess_messages'] = (Total_Fusion['message_count'] - Total_Fusion['messages_included']).clip(lower=0)
Total_Fusion['excess_gb'] = (Total_Fusion['gb_used'] - (Total_Fusion['mb_per_month_included'] / 1024)).clip(lower=0)

Total_Fusion['extra_minutes_cost'] = Total_Fusion['excess_minutes'] * Total_Fusion['usd_per_minute']
Total_Fusion['extra_messages_cost'] = Total_Fusion['excess_messages'] * Total_Fusion['usd_per_message']
Total_Fusion['extra_gb_cost'] = Total_Fusion['excess_gb'] * Total_Fusion['usd_per_gb']
Total_Fusion['monthly_revenue'] = (Total_Fusion['usd_monthly_pay'] + 
                                   Total_Fusion['extra_minutes_cost'] + 
                                   Total_Fusion['extra_messages_cost'] + 
                                   Total_Fusion['extra_gb_cost'])
print("Ingresos calculados correctamente:")
print(Total_Fusion[['user_id', 'year_month', 'plan', 'monthly_revenue']].head())
```

```python


# Prueba las hipótesis
#Identificar usuarios del área NY-NJ
Total_Fusion_with_city = pd.merge(Total_Fusion, users[['user_id', 'city']], on='user_id', how='left')

# Crear variable para identificar región NY-NJ
Total_Fusion_with_city['is_ny_nj'] = Total_Fusion_with_city['city'].str.contains('NY-NJ', na=False)

# Separar ingresos por región
NY_NJ_revenue = Total_Fusion_with_city[Total_Fusion_with_city['is_ny_nj'] == True]['monthly_revenue']
other_regions_revenue = Total_Fusion_with_city[Total_Fusion_with_city['is_ny_nj'] == False]['monthly_revenue']


print(f"Usuarios NY-NJ: {len(NY_NJ_revenue)}")
print(f"Usuarios otras regiones: {len(other_regions_revenue)}")



```

Aunque filtraste correctamente el dataset para crear los conjuntos de datos, debes implemetar las siguiente pruebas de hipótesis:

    - El ingreso promedio de los usuarios de las tarifas Ultimate y Surf difiere.
    - El ingreso promedio de los usuarios en el área de estados Nueva York-Nueva Jersey es diferente al de los usuarios de otras regiones.



Para comparar distribuciones de manera adecuada, es importante seguir el siguiente proceso:

1. Verificar si los datos siguen una distribución normal, lo cual se puede hacer con pruebas como **Shapiro-Wilk** o **Kolmogorov-Smirnov**.
2. Si los datos son normales, el siguiente paso es comprobar si las varianzas son iguales mediante pruebas como **Levene**.

Estos dos factores (normalidad y homogeneidad de varianzas) determinan la prueba estadística que debes usar: si se cumplen ambos, puedes aplicar una prueba paramétrica como el **t-tes**t; si no, debes optar por una prueba no paramétrica como **Mann-Whitney U**. Este enfoque te permite elegir la herramienta más adecuada para que tus conclusiones sean fiables y válidas.


</div>

```python
from scipy import stats

# Separar los ingresos por tipo de plan
ultimate_revenue = Total_Fusion_with_city[Total_Fusion_with_city['plan'] == 'ultimate']['monthly_revenue']
surf_revenue = Total_Fusion_with_city[Total_Fusion_with_city['plan'] == 'surf']['monthly_revenue']
t_stat, p_value = stats.ttest_ind(ultimate_revenue, surf_revenue, equal_var=False)
print("Prueba de hipótesis: Ingreso promedio entre Ultimate y Surf")
print(f"Estadístico t: {t_stat:.3f}")
print(f"Valor p: {p_value:.3f}")

# Evaluar hipótesis
alpha = 0.05
if p_value < alpha:
    print(" Se rechaza la hipótesis nula → Los ingresos promedio son diferentes.")
else:
    print(" No se rechaza la hipótesis nula → No hay diferencia significativa en los ingresos promedio.")
```
